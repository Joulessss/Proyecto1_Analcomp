```python
df.columns 
```

variable adicional

```python
df['segmento'] = df['cole_naturaleza'] + '_' + df['cole_area_ubicacion']
```

## Descriptivas

```python
df.describe(include='all')
```

```python
#skewness
scores = ['punt_sociales_ciudadanas', 'punt_matematicas', 'punt_ingles',
          'punt_c_naturales', 'punt_lectura_critica', 'punt_global']
print(df[scores].skew().round(3).to_string())
```

```python
# Distribución de variables categóricas
print(df['cole_naturaleza'].value_counts())
print(df['cole_area_ubicacion'].value_counts())
print(df['cole_caracter'].value_counts())
print(df['estu_privado_libertad'].value_counts())

# Tabla resumen por segmentos clave
print("descripción por segmento")
print(df.groupby('segmento')['punt_sociales_ciudadanas']
      .agg(['mean','median','std','count']).round(2).sort_values('mean'))
```
