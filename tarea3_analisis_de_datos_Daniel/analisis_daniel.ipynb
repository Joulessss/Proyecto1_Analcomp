## Boxplots por segmento

```python
fig = make_subplots(
    rows=1, cols=2,
    subplot_titles=['Por Naturaleza del Colegio', 'Por Área de Ubicación'],
    horizontal_spacing=0.12,
)

for nombre, filtro in [('Oficial (Público)',   df['cole_naturaleza']=='OFICIAL'),
                        ('No Oficial (Privado)', df['cole_naturaleza']=='NO OFICIAL')]:
    fig.add_trace(go.Box(
        y=df[filtro]['punt_sociales_ciudadanas'],
        name=nombre,
        boxmean='sd',
        hovertemplate=(
            f'<b>{nombre}</b><br>'
            'Mediana: %{median}<br>'
            'Q1: %{q1}<br>'
            'Q3: %{q3}<br>'
            'μ: %{mean:.1f}<extra></extra>'
        ),
    ), row=1, col=1)

for nombre, filtro in [('Urbano', df['cole_area_ubicacion']=='URBANO'),
                        ('Rural',  df['cole_area_ubicacion']=='RURAL')]:
    fig.add_trace(go.Box(
        y=df[filtro]['punt_sociales_ciudadanas'],
        name=nombre,
        boxmean='sd',
        hovertemplate=(
            f'<b>{nombre}</b><br>'
            'Mediana: %{median}<br>'
            'Q1: %{q1}<br>'
            'Q3: %{q3}<br>'
            'μ: %{mean:.1f}<extra></extra>'
        ),
    ), row=1, col=2)

fig.update_layout(
    title=dict(text='Diagramas de Caja — Ciencias Sociales por Segmento', x=0.5),
    height=500,
    paper_bgcolor='white',
    plot_bgcolor='white',
)
fig.update_yaxes(title_text='Puntaje C. Sociales', gridcolor='lightgrey', linecolor='lightgrey')
fig.update_xaxes(linecolor='lightgrey')

fig.show()

# ── Tests de significancia estadística ────────────────────────────────────
for label, g1, g2 in [
    ('Oficial vs No Oficial', df[df['cole_naturaleza']=='OFICIAL']['punt_sociales_ciudadanas'],
                              df[df['cole_naturaleza']=='NO OFICIAL']['punt_sociales_ciudadanas']),
    ('Urbano vs Rural',       df[df['cole_area_ubicacion']=='URBANO']['punt_sociales_ciudadanas'],
                              df[df['cole_area_ubicacion']=='RURAL']['punt_sociales_ciudadanas']),
]:
    t, p = stats.ttest_ind(g1, g2)
    print(f"Prueba t  {label:<25}: t={t:.2f},  p={p:.2e}")

# Para Dash: dcc.Graph(figure=fig)
```

## Gráficos de violÍn

```python
fig = make_subplots(
    rows=1, cols=2,
    subplot_titles=[
        'Por Segmento ',
        'Privados de Libertad vs Población General'
    ],
    horizontal_spacing=0.12,
)

# ── Panel 1: 4 segmentos ───────────────────────────────────────────────────
for segmento in ['NO OFICIAL_RURAL', 'NO OFICIAL_URBANO', 'OFICIAL_RURAL', 'OFICIAL_URBANO']:
    fig.add_trace(go.Violin(
        y=df[df['segmento']==segmento]['punt_sociales_ciudadanas'],
        name=segmento,
        box_visible=True,
        meanline_visible=True,
        hovertemplate=f'<b>{segmento}</b><br>Puntaje: %{{y}}<extra></extra>',
    ), row=1, col=1)

# ── Panel 2: Privados de Libertad vs población general ────────────────────
for nombre, filtro in [
    ('NO', df['estu_privado_libertad']=='N'),
    ('SI', df['estu_privado_libertad']=='S'),
]:
    fig.add_trace(go.Violin(
        y=df[filtro]['punt_sociales_ciudadanas'],
        name=nombre,
        box_visible=True,
        meanline_visible=True,
        hovertemplate=f'<b>{nombre}</b><br>Puntaje: %{{y}}<extra></extra>',
        showlegend=False,
    ), row=1, col=2)

fig.update_layout(
    title=dict(text='Diagramas de Violín — Ciencias Sociales por Segmento', x=0.5),
    height=500,
    width=1200,
    paper_bgcolor='white',
    plot_bgcolor='white',
)
fig.update_yaxes(title_text='Puntaje C. Sociales', gridcolor='lightgrey', linecolor='lightgrey')
fig.update_xaxes(linecolor='lightgrey', tickangle=-30)

fig.show()
# Para Dash: dcc.Graph(figure=fig))
```

# Matriz de Correlación

```python
corr = df[scores].corr().round(3)

fig = make_subplots(
    rows=1, cols=2,
    subplot_titles=[
        'Matriz de Correlación — Todos los Puntajes',
        'Correlación con Ciencias Sociales'
    ],
    horizontal_spacing=0.15,
    column_widths=[0.6, 0.4],
)

# ── Panel 1: Heatmap matriz de correlación ─────────────────────────────────
fig.add_trace(go.Heatmap(
    z=corr.values,
    x=scores_short,
    y=scores_short,
    colorscale='RdYlGn',
    zmin=0, zmax=1,
    text=corr.values,
    texttemplate='%{text:.2f}',
    textfont=dict(size=10),
    colorbar=dict(x=0.55, len=0.9, thickness=12),
    hovertemplate='%{y} — %{x}<br>r = %{z:.3f}<extra></extra>',
), row=1, col=1)

# ── Panel 2: Barras horizontales correlación con C. Sociales ───────────────
corr_sc = (corr['punt_sociales_ciudadanas']
           .drop('punt_sociales_ciudadanas')
           .sort_values(ascending=True))

fig.add_trace(go.Bar(
    x=corr_sc.values,
    y=scores_short[:5],
    orientation='h',
    text=[f'{v:.3f}' for v in corr_sc.values],
    textposition='outside',
    hovertemplate='%{y}<br>r = %{x:.3f}<extra></extra>',
    showlegend=False,
), row=1, col=2)

fig.update_layout(
    title=dict(text='Correlaciones entre Puntajes Saber 11', x=0.5),
    height=450,
    width=1100,
    paper_bgcolor='white',
    plot_bgcolor='white',
)
fig.update_xaxes(gridcolor='lightgrey', linecolor='lightgrey', tickangle=-30)
fig.update_yaxes(gridcolor='lightgrey', linecolor='lightgrey')
fig.update_xaxes(title_text='Coeficiente de Pearson', row=1, col=2)

fig.show()

# Interpretación automática
print("Correlaciones con Ciencias Sociales (de mayor a menor):")
for materia, val in corr_sc.sort_values(ascending=False).items():
    materia_short = scores_short[scores.index(materia)]
    print(f"  {materia_short:<20}: r = {val:.3f}")
# Para Dash: dcc.Graph(figure=fig)
```
